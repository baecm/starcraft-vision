```python
FPS = 24
def format_mm_ss_ms(frame, fps):
    total_seconds = frame / fps
    minutes = int(total_seconds // 60)
    seconds = total_seconds % 60
    # :05.2f -> 전체 5자리(00.00), 소수점 아래 2자리, 남는 앞자리는 0으로 채움
    return f"{minutes:02d}:{seconds:05.2f}"
```

```python
df['time'] = df['frame'].apply(lambda x: format_mm_ss_ms(x, FPS))
```
