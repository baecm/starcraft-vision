```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

cmap = plt.get_cmap("Set2")

color_vanilla = cmap(0)
color_kbrs = cmap(1)
# 1. 비교하고 싶은 metric 선택
metrics = ["ic@000", "ic@030", "ic@050", "ic_ratio", "median_ir"]
metrics_for_ticks = {
    "ic@000": "intersection@any",
    "ic@030": "intersection@0.3",
    "ic@050": "intersection@0.5",
    "ic_ratio": "intersection\nratio",
    "median_ir": "median\nintersection\nratio"
}
# 필요에 따라 mean_density, mean_mixture 등 추가해도 됨

# 2. mean / std 추출 (row: describe index, col: metric)
vanilla_means = desc_vanilla_all.loc["mean", metrics]
vanilla_stds  = desc_vanilla_all.loc["std", metrics]

kbrs_means = desc_kbrs_all.loc["mean", metrics]
kbrs_stds  = desc_kbrs_all.loc["std", metrics]

# 3. bar chart: metric 별로 vanilla vs kbrs 나란히
x = np.arange(len(metrics))
width = 0.35

plt.figure(figsize=(10, 5), dpi=300)

plt.bar(
    x - width / 2,
    vanilla_means.values,
    width,
    yerr=vanilla_stds.values,
    capsize=4,
    label="Mask R-CNN",
    color=color_vanilla,
)

plt.bar(
    x + width / 2,
    kbrs_means.values,
    width,
    yerr=kbrs_stds.values,
    capsize=4,
    label="Mask R-CNN w/ KBRS",
    color=color_kbrs,
)

tick_labels = [metrics_for_ticks[m] for m in metrics]
plt.xticks(x, tick_labels, rotation=0)
plt.ylabel("score")
plt.title("Mask R-CNN vs Mask R-CNN w/ KBRS (mean ± std over all runs)")
plt.legend(loc="lower right")
# plt.tight_layout()
plt.show()
```

```python
def collect_keys_by_level(obj, level=0, result=None):
    """
    중첩된 dict (그리고 dict가 들어 있는 list)에서
    level별로 key를 수집해서 {level: set(keys)} 형태로 반환.
    """
    if result is None:
        result = {}

    if isinstance(obj, dict):
        # 현재 level에 key들 추가
        result.setdefault(level, set())
        for k, v in obj.items():
            result[level].add(k)
            # value가 dict나 list이면 더 내려가서 탐색
            if isinstance(v, (dict, list)):
                collect_keys_by_level(v, level + 1, result)

    elif isinstance(obj, list):
        # 리스트 안에 dict가 있을 수 있으므로 처리
        for item in obj:
            if isinstance(item, (dict, list)):
                collect_keys_by_level(item, level, result)

    return result
```
