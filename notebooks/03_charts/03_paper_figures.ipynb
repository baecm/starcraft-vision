```python
def collect_keys_by_level(obj, level=0, result=None):
    """
    중첩된 dict (그리고 dict가 들어 있는 list)에서
    level별로 key를 수집해서 {level: set(keys)} 형태로 반환.
    """
    if result is None:
        result = {}

    if isinstance(obj, dict):
        # 현재 level에 key들 추가
        result.setdefault(level, set())
        for k, v in obj.items():
            result[level].add(k)
            # value가 dict나 list이면 더 내려가서 탐색
            if isinstance(v, (dict, list)):
                collect_keys_by_level(v, level + 1, result)

    elif isinstance(obj, list):
        # 리스트 안에 dict가 있을 수 있으므로 처리
        for item in obj:
            if isinstance(item, (dict, list)):
                collect_keys_by_level(item, level, result)

    return result
```

```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_ablation_group(df: pd.DataFrame, group_name: str):
    # df columns: setting, @any, @0.3, @0.5, multiple, ratio
    df = df.sort_values("setting").reset_index(drop=True)

    x = df["setting"].to_numpy(float)
    xlabels = [str(v) for v in df["setting"].tolist()]

    # -------------------------
    # Figure 1: @any/@0.3/@0.5 together
    # -------------------------
    fig, ax = plt.subplots(figsize=(8.5, 4.2), dpi=300)
    ax.plot(x, df["@any"], marker="o", label="@any")
    ax.plot(x, df["@0.3"], marker="o", label="@0.3")
    ax.plot(x, df["@0.5"], marker="o", label="@0.5")
    ax.set_title(f"{group_name}: intersection@tau")
    ax.set_xlabel(f"{group_name} setting")
    ax.set_ylabel("score")
    ax.set_xticks(x)
    ax.set_xticklabels(xlabels)
    ax.grid(True, linewidth=0.5, alpha=0.4)
    ax.legend()
    # ax.set_ylim(0.7, 0.85)
    fig.tight_layout()
    plt.show()

    # -------------------------
    # Figure 2: multiple + ratio (separate panels)
    # -------------------------
    fig, (ax1, ax2) = plt.subplots(nrows=2, ncols=1, figsize=(8.5, 6.2), sharex=True)

    ax1.plot(x, df["multiple"], marker="o")
    ax1.set_title(f"{group_name}: multiple")
    ax1.set_ylabel("multiple")
    ax1.grid(True, linewidth=0.5, alpha=0.4)

    ax2.plot(x, df["ratio"], marker="o")
    ax2.set_title(f"{group_name}: ratio")
    ax2.set_xlabel(f"{group_name} setting")
    ax2.set_ylabel("ratio")
    ax2.set_xticks(x)
    ax2.set_xticklabels(xlabels)
    ax2.grid(True, linewidth=0.5, alpha=0.4)
    # ax2.set_ylim(0.65, 0.75)

    fig.tight_layout()
    plt.show()
```

```python
df_mixture = pd.DataFrame(
    [{"setting": k, **dict(zip(metrics, v))} for k, v in groups["mixture"].items()]
)

plot_ablation_group(df_mixture, "mixture")
```

```python
df_density = pd.DataFrame(
    [{"setting": k, **dict(zip(metrics, v))} for k, v in groups["density"].items()]
)

plot_ablation_group(df_density, "density")
```
