```python
import pandas as pd
import numpy as np
from scipy.spatial.distance import cdist

def track_streaks(df, dist_threshold=10.0):
    # 1. 준비: x, y 추출 및 이미지 순서 정렬
    df['x'] = df['bbox'].apply(lambda b: b[0])
    df['y'] = df['bbox'].apply(lambda b: b[1])
    image_ids = sorted(df['image_id'].unique())
    
    # 결과 저장을 위한 컬럼
    df['streak_id'] = -1
    next_streak_id = 0
    
    # 첫 프레임 객체들에 ID 부여
    first_frame_idx = df[df['image_id'] == image_ids[0]].index
    for idx in first_frame_idx:
        df.at[idx, 'streak_id'] = next_streak_id
        next_streak_id += 1
        
    # 2. 프레임 순회하며 매칭
    for i in range(len(image_ids) - 1):
        curr_id = image_ids[i]
        next_id = image_ids[i+1]
        
        curr_rows = df[df['image_id'] == curr_id]
        next_rows = df[df['image_id'] == next_id]
        
        if curr_rows.empty or next_rows.empty:
            continue
            
        # 두 프레임 간의 모든 거리 계산 (Euclidean)
        curr_coords = curr_rows[['x', 'y']].values
        next_coords = next_rows[['x', 'y']].values
        dist_matrix = cdist(curr_coords, next_coords, metric='euclidean')
        
        # Greedy Matching
        matched_next_indices = []
        for curr_idx_in_matrix in range(len(curr_rows)):
            # 현재 객체에서 가장 가까운 다음 프레임 객체 찾기
            min_dist = dist_matrix[curr_idx_in_matrix].min()
            next_idx_in_matrix = dist_matrix[curr_idx_in_matrix].argmin()
            
            if min_dist < dist_threshold and next_idx_in_matrix not in matched_next_indices:
                # 매칭 성공: 이전 streak_id 계승
                actual_next_idx = next_rows.index[next_idx_in_matrix]
                actual_curr_idx = curr_rows.index[curr_idx_in_matrix]
                
                df.at[actual_next_idx, 'streak_id'] = df.at[actual_curr_idx, 'streak_id']
                matched_next_indices.append(next_idx_in_matrix)
        
        # 매칭되지 않은 다음 프레임 객체들은 새로운 Streak ID 부여
        for j, next_idx in enumerate(next_rows.index):
            if j not in matched_next_indices:
                df.at[next_idx, 'streak_id'] = next_streak_id
                next_streak_id += 1
                
    return df
```

```python
# 실행
df_vanilla_tracked = track_streaks(df_vanilla, dist_threshold=15.0)

# Streak 길이(지속성) 측정
streak_lengths = df_vanilla_tracked.groupby('streak_id').size()
print(streak_lengths)
```

```python
# 1. 각 Streak ID별로 포함된 데이터(프레임) 개수 카운트
vanilla_streak_counts = df_vanilla_tracked.groupby('streak_id').size()

# 2. 전체 Streak의 평균 지속 프레임 수 계산
average_vanilla_streak_length = vanilla_streak_counts.mean()

print(f"평균 Streak 지속 프레임: {average_vanilla_streak_length:.2f} frames")
```

```python
long_vanilla_streaks = vanilla_streak_counts[vanilla_streak_counts >= 24]
avg_long_vanilla_streak = long_vanilla_streaks.mean()

print(f"24프레임 이상 유지된 Streak의 평균: {avg_long_vanilla_streak:.2f} frames")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from ipywidgets import interact, IntSlider

def interactive_rank_viewer(df_tracked, grid_size=128):
    # 1. 프레임별로 Score 순위(Rank) 계산 (내림차순: 점수 높은 것이 1등)
    # method='first'를 사용하면 점수가 같더라도 중복 순위 없이 1, 2등을 나눕니다.
    df_tracked['rank'] = df_tracked.groupby('image_id')['score'].rank(ascending=False, method='first').astype(int)
    
    frame_list = sorted(df_tracked['image_id'].unique())
    
    # Streak ID별 고유 색상은 유지 (객체 추적 연속성 확인용)
    all_streak_ids = df_tracked['streak_id'].unique()
    cmap = plt.cm.get_cmap('tab20', 20)
    streak_colors = {sid: cmap(i % 20) for i, sid in enumerate(all_streak_ids)}
    streak_colors[-1] = (0.5, 0.5, 0.5, 1)

    def plot_frame(frame_id):
        fig, ax = plt.subplots(figsize=(10, 10))
        
        ax.set_xlim(0, grid_size)
        ax.set_ylim(0, grid_size)
        ax.invert_yaxis()
        ax.set_aspect('equal')
        ax.grid(True, linestyle=':', alpha=0.5)
        
        now_df = df_tracked[df_tracked['image_id'] == frame_id].copy()
    
        # [핵심] Rank를 기준으로 내림차순 정렬 (예: 5위 -> 4위 -> ... -> 1위 순서로 그리기)
        # 이렇게 하면 1위(Rank 1)가 가장 마지막에 그려져서 최상단에 위치하게 됩니다.
        now_df = now_df.sort_values(by='rank', ascending=False)
    
        for _, row in now_df.iterrows():
            sid = row['streak_id']
            score = row['score']
            rank = row['rank']
            color = streak_colors[sid]
            
            # 1위인 경우 테두리를 더 굵게 표현하여 강조할 수도 있습니다.
            edge_width = 3 if rank == 1 else 1.5
            
            rect = patches.Rectangle(
                (row['x'], row['y']), 20, 12,
                linewidth=edge_width,
                edgecolor=color,
                facecolor=color,
                alpha=0.4, # 겹침 확인을 위해 투명도 유지
                zorder=3 + (100 / rank) # zorder를 이용해 명시적으로 레이어 순서 지정 가능
            )
            ax.add_patch(rect)
            
            # 라벨 표시
            label_text = f"Rank: {rank}\n{score:.4f}"
            ax.text(
                row['x'], row['y'] - 2, 
                label_text, 
                color='white', fontsize=8, fontweight='bold',
                bbox=dict(facecolor=color, alpha=0.8, edgecolor='none', pad=2),
                zorder=4 + (100 / rank) # 텍스트도 사각형 위에 오도록 zorder 설정
            )
    
        plt.show()

    interact(plot_frame, 
             frame_id=IntSlider(min=min(frame_list), max=max(frame_list), 
                                step=1, value=min(frame_list),
                                description='Frame:'))

# 실행 전 데이터프레임에 x, y가 있는지 확인하세요!
interactive_rank_viewer(df_vanilla_tracked)
```
